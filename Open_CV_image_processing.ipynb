# Implementation OPEN CV for image processing


```python
# importing libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
#Add images path 
image_path='Sample_cv.jpeg'
```

```python
#load the image
img = cv2.imread(image_path)
```

```python
#conert to RGB and grayscale
img_rgb=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_gray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
#Display Image
#display bath version side by side 
plt.figure(figsize=(12,12))
plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis("off")
plt.tight_layout()
plt.show()
```
