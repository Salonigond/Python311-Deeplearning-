# Implementation of RNN for Text classification


```python
# import libraries

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#prepare the text data
texts = [
    "I love this movie", 
    "This film is great", 
    "I hate this movie", 
    "This film is bad"
]
labels = [1,1,0,0] #1 for positive 0 for negative
```

```python
# tokenization 
tokenizer= Tokenizer()
tokenizer.fit_on_texts(texts)
sequences =tokenizer.texts_to_sequences(texts)
#padding sequences
X=pad_sequences(sequences, maxlen=5)
y= np.array(labels)
```

```python
#Build RNN model
model = Sequential()
model.add(Embedding (input_dim=100, output_dim=8, input_length=5))
model.add(SimpleRNN(10, activation='relu'))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
#compile the model
model.compile(optimizer='adam', loss= 'binary_crossentropy', metrics=['accuracy'])
```

```python
#train the modedl
model.fit(X,y,epochs=100,verbose=0)
```

```python
#test the model
test_text=['I love this flim']
test_seq=tokenizer.texts_to_sequences(test_text)
test_pad=pad_sequences(test_seq, maxlen=5)

#model prediction
prediction= model.predict(test_pad)
print("Prediction:", prediction)
```
