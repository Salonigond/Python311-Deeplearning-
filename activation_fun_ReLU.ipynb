```python
# Import necessary library
import numpy as np
```

```python
# Define the ReLU function
def relu(x):
    return np.maximum(0, x)
```

```python
# Example usage of ReLU function
x1 = 2
print("ReLU({}) = {}".format(x1, relu(x1)))
x2 = -1
print("ReLU({}) = {}".format(x2, relu(x2)))
x3 = 0
print("ReLU({}) = {}".format(x3, relu(x3)))
```
