## Implement sigmoid function using Numpy

```python
# Import labraries
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
X=np.linspace(-10,10,100)
Z=1/(1+np.exp(-X))
plt.plot(X,Z)
plt.xlabel("X")
plt.ylabel("Sigmoid(X)")
plt.show()
```
