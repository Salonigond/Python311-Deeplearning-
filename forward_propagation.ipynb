## Implementation of Forward Propagation


```python
#import libraries
import numpy as np
import pandas as pd
```

```python
#create sample dataset
data={'cgpa':[8.5, 9.2, 7.8], 'profile_score':[85,92,78], 'lpa':[10,12,8]}
df=pd.DataFrame(data)
X=df[['cgpa','profile_score']].values
print(X)
```

```python
#Initialize parameters
def init_parameter():
    np.random.seed(1)
    W=np.random.randn(2,1)*0.01
    b=np.zeros((1,1))
    return W,b
```

```python
#Define Forward Propagation
#compute Z=WX +b
#sigmodid activition ensure values b/w 0 and 1
def forward_propagation(X,W,b):
    Z=np.dot(X,W)+b
    A=1/(1+np.exp(-Z))
    return A
```

```python
#Exexute Forward Propagation
W,b= init_parameter()
A=forward_propagation(X,W,b)
print("Final Output:",A)
```
