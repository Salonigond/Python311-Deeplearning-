# Implementation of Lemmatization text using NLTK


```python
# import libraries
import nltk
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('owm-1.4')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
#Lemmatization Text with NLTK 
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
text = "The cats were running faster than the dogs"
tokens = word_tokenize(text)
lemmatized_words= [lemmatizer.lemmatize(word) for word in tokens]
print(f"original word:{text}")
print(f"Lemmatize word:{lemmatized_words}")
```
