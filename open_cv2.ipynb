# Python program to illustrate
# simple thresholding type on an image

```python
# organizing imports
import cv2
import numpy as np
```

```python
# path to input image is specified and
# image is loaded with imread command
image1 = cv2.imread('input1.jpg')
```

```python
# cv2.cvtColor is applied over the
# image input with applied parameters
# to convert the image in grayscale
img = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
```

```python
# applying different thresholding
# techniques on the input image
# all pixels value above 120 will
# be set to 255
ret, thresh1 = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY)
ret, thresh2 = cv2.threshold(img, 120, 255, cv2.THRESH_BINARY_INV)
ret, thresh3 = cv2.threshold(img, 120, 255, cv2.THRESH_TRUNC)
ret, thresh4 = cv2.threshold(img, 120, 255, cv2.THRESH_TOZERO)
ret, thresh5 = cv2.threshold(img, 120, 255, cv2.THRESH_TOZERO_INV)
```

```python
# the window showing output images
# with the corresponding thresholding
# techniques applied to the input images
cv2.imshow('Binary Threshold', thresh1)
cv2.imshow('Binary Threshold Inverted', thresh2)
cv2.imshow('Truncated Threshold', thresh3)
cv2.imshow('Set to 0', thresh4)
cv2.imshow('Set to 0 Inverted', thresh5)
```

```python
# De-allocate any associated memory usage
if cv2.waitKey(0) & 0xff == 27:
    cv2.destroyAllWindows()
```

## Adaptive Thresholding

### Python program to illustrate
### Otsu thresholding type on an image


```python
# organizing imports
import cv2
import numpy as np
```

```python
# path to input image is specified and
# image is loaded with imread command
image1 = cv2.imread('input1.jpg')
```

```python
# cv2.cvtColor is applied over the
# image input with applied parameters
# to convert the image in grayscale
img = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
```

```python
# applying Otsu thresholding
# as an extra flag in binary
# thresholding
thresh1 = cv2.adaptiveThreshold(img, 120, 255, cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY, 199, 5)
thresh2 = cv2.adaptiveThreshold(img, 120, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY, 199, 5)
```

```python
# the window showing output image
# with the corresponding thresholding
# techniques applied to the input image
cv2.imshow('Adaptive Mean', thresh1)
cv2.imshow('Adaptive Gaussian', thresh2)
```

```python
# De-allocate any associated memory usage
if cv2.waitKey(0) & 0xff == 27:
    cv2.destroyAllWindows()
```

## structuring element

```python
import cv2
```

```python
# Load an image
img = cv2.imread('image.jpg', 0)  # Load in grayscale
```

```python
# Create a rectangular structuring element of size 18x18
rect_kernel = 
cv2.getStructuringElement(cv2.MORPH_RECT, (18, 18))
```

```python
# Apply morphological operations (e.g., dilation) using 
the structuring element
dilated_img = cv2.dilate(img, rect_kernel, iterations=1)
```

```python
# Show the dilated image
cv2.imshow('Dilated Image', dilated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Dilation operation

```python
import cv2
```

```python
# Load an image in grayscale
img = cv2.imread('image.jpg', 0)
```

```python
# Apply thresholding to get a binary image (using Otsu 
thresholding for example)
ret, thresh1 = cv2.threshold(img, 0, 255, cv2.THRESH_OTSU | 
cv2.THRESH_BINARY_INV)
```

```python
# Create a rectangular structuring element (18x18)
rect_kernel = cv2.getStructuringElement(cv2.MORPH_RECT, 
(18, 18))
```

```python
# Apply dilation to the thresholded image using the kernel
dilated_img = cv2.dilate(thresh1, rect_kernel, iterations=1)
```

```python
# Show the dilated image
cv2.imshow('Dilated Image', dilated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## cv2.boundingRect(cnt)

```python
import cv2
import numpy as np
```

```python
# Load the image and convert it to grayscale
image = cv2.imread('image.png')
gray= cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
# Threshold the image to get a binary image
_, thresholded= cv2.threshold(gray, 127, 255, 
cv2.THRESH_BINARY)
```

```python
# Find contours
contours, _ = cv2.findContours(thresholded, 
cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
```

```python
# Loop through each contour
for cntin contours:
# Get the bounding rectangle for each contour
x, y, w, h = cv2.boundingRect(cnt)
```

```python
# Show the image with bounding rectangles
cv2.imshow('Bounding Rectangles', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpyas np
```

```python
# Load the image
image = cv2.imread('image.png')
```

```python
# Convert it to grayscale
gray= cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
# Apply thresholding to get a binary image
_, thresholded= cv2.threshold(gray, 127, 255, 
cv2.THRESH_BINARY)
```

```python
# Find contours
contours, _ = cv2.findContours(thresholded, 
cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
```

```python
# Loop through each contour
for cntin contours:
# Get the bounding rectangle for each contour
x, y, w, h = cv2.boundingRect(cnt)
```

```python
# Crop the region of interest (ROI) from the image
cropped = image[y:y+ h, x:x + w]
```

```python
# You can now save or process the cropped region, for example:
cv2.imshow("Cropped", cropped)
cv2.waitKey(0)
```

```python
# Optionally, display the original image with rectangles (to see 
where the cropping was done)
for cntin contours:
x, y, w, h = cv2.boundingRect(cnt)
cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)
```

```python
cv2.imshow('Image with Rectangles', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
