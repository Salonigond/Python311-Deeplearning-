# Implementation of punctuation removal


```python
#impport libraries
import re
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
```

```python
text ="This is a sample sentence, showing off stop word filteration"
tokens = word_tokenize(text)
#regular expression to match punctuation
clean_tokens=[re.sub(r'[^\w\s]','',tokens) for token in tokens if re.sub(r'[^\w\s]','',token)]
print(clean_tokens)
```
